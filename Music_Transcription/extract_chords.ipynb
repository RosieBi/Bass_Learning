```python
import librosa
from numpy.fft import fft, fftfreq
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Audio
```

```python
input_filename = "mp3_input/bass_NGHYB_clean.mp3"
audio, sr = librosa.load(input_filename)
```

```python
n = len(audio)
duration = n/sr
duration
```

```python
Audio(input_filename)
```

```python
###Discrete Fourier Transformation
audio_fft = fft(audio)
```

```python
amp_spectrum = np.abs(audio_fft)
amp_spectrum = amp_spectrum/np.max(amp_spectrum)
freqs = fftfreq(n, 1 / (sr))

plt.plot(freqs[:500], amp_spectrum[:500])
plt.xlabel("Frequency [Hz]")
plt.ylabel("Normalized Amplitude")
plt.title("Amplitude Spectrum of the Signal")
plt.show()
```
